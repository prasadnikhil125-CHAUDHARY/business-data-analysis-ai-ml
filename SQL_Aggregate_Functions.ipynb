```python
def sql(query):
  return pd.read_sql(query, conn)
```

```python
sql("SELECT * FROM EMPLOYEE")
```

```python
# Retriev selected columns
sql("SELECT Name, Department FROM Employee")
```

```python
# Select salary >60000
sql("SELECT * FROM Employee WHERE Salary >60000")
```

```python
# Employee from Delhi
sql("SELECT * FROM Employee WHERE City = 'Delhi'")
```

```python
# Between age 20 and 25
sql("SELECT * FROM Employee WHERE Age BETWEEN 20 AND 25")
```

```python
# Highest Salary
sql("SELECT * FROM Employee ORDER BY Salary DESC")
```

```python
# Count
sql("SELECT COUNT(*) FROM Employee")
```

```python
# Average
sql("SELECT AVG(Salary) FROM Employee")
```

```python
# Max
sql("SELECT MAX(Salary) FROM Employee")
```

```python
# Min age
sql("SELECT MIN(Age) FROM Employee")
```

```python
# Total salary budget
sql("SELECT SUM(Salary) FROM Employee")
```

```python
# Dept wise Avg Salary
sql("SELECT Department, AVG(Salary) FROM Employee GROUP BY Department")
```

```python
# Dept wise employee count
sql("SELECT Department, COUNT(*) FROM Employee GROUP BY Department")
```

```python
# Dept wise highest salary
sql("SELECT Department, MAX(Salary) AS MaxSalary FROM Employee GROUP BY Department HAVING MaxSalary>65000")
```
