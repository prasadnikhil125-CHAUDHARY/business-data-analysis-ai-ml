# **Supply Chain Analytics**

```python
# Upload data
import sqlite3
import requests
import pandas as pd

url = "https://github.com/YBIFoundation/SQL/raw/main/SupplyChainAdvanceAnalytics.db"

response = requests.get(url)
with open("SupplyChainAdvanceAnalytics.db", "wb") as f:
    f.write(response.content)

conn = sqlite3.connect("SupplyChainAdvanceAnalytics.db")

def sql(query):
    return pd.read_sql(query, conn)
```

```python
sql("SELECT * FROM supply_chain")
```

```python
# Total records
sql("SELECT COUNT(*) FROM supply_chain")
```

```python
sql("SELECT Product,	Category FROM supply_chain")
```

```python
# Electronics
sql("SELECT * FROM supply_chain WHERE Category = 'Electronics'")
```

```python
# Net_Sales
sql("SELECT * FROM supply_chain WHERE Net_Sales>200000;")
```

```python
# Electronics AND >200000
sql("SELECT * FROM supply_chain WHERE Category = 'Electronics' AND Net_Sales>200000")
```

```python
# Text Functions- LOWER, UPPER, TRIM
sql("SELECT Warehouse, TRIM(Warehouse) FROM supply_chain ")
```

```python
# Text - LENGTH
sql("SELECT Category, Product, LENGTH(Product) FROM supply_chain")
```

```python
# REPLACE - Warehouse
sql("SELECT Warehouse, REPLACE(Warehouse,'Chennai','Madras') FROM supply_chain")
```

```python
# Substring
sql("SELECT Product, SUBSTR(Product,1,3) AS Product_ID FROM supply_chain")
```

```python
sql("SELECT * FROM supply_chain")
```

```python
# Extract Year
sql("SELECT strftime('%Y', Order_Date) AS Order_Year FROM supply_chain")
```

```python
# Extract Month
sql("SELECT strftime('%m', Order_Date) AS Order_Month FROM supply_chain")
```

```python
# Extract date
sql("SELECT strftime('%d', Order_Date) AS Order_Date FROM supply_chain")
```
